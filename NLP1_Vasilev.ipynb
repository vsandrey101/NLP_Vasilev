# Загрузка модели

Будет установлена QA модель (ru-версия) для ответов на вопросы исходя из предоставленного контекста

```python
from deeppavlov import build_model
model = build_model('squad_ru_bert', download=True, install=True)
```

# Демонстрация модели

Модели, на основе данных фрагментов текста, необходимо будет ответить на вопросы

```python
text = "При прокладке силовых линий применяют двух- и трехжильные силовые кабели с резиновой изоляцией и с сечением медных жил 1; 1,5 и 2,5 мм2 , а алюминиевых — 2,5 и 4 мм2."
question = "Какое сечение у алюминиевых силовых кабелей?"
model([text], [question])
```

```python
text = "Приемник электрической энергии (электроприемник) - аппарат, агрегат и др., предназначенный для преобразования электрической энергии в другой вид энергии."
question = "Для чего нужен электроприемник?"
model([text], [question])
```

```python
text = "В течение двух месяцев исследования 4 группы студентов (110 человек) просмотрели свои результаты более 10 тысяч раз, что свидетельствует о высокой степени вовлечённости обучающихся и активном использовании данных сгенерированных диаграмм для анализа собственных результатов."
question = "Сколько человек участвовало в исследовании?"
model([text], [question])
```
