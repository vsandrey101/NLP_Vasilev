# Чат с локальной моделью

```python
import ollama
#наличие установленных моделей
models = ollama.list()
print(models)
```

```python
from datetime import datetime
#телеграм заблокировали, будем чатиться в блокноте :)
#модель ограниченная -> лучше общаться на английском (больше обучена на нем)
def chat():
    history = []
    while True:
        user_in = input(f"{datetime.now():%H:%M:%S} Юзер: ")
        if user_in == "EXIT":
            break
        history.append({'role': 'user', 'content': user_in})
        response = ollama.chat(
            model="deepseek-r1:1.5b",
            messages=history,
            stream=False # Устанавливаем False для получения полного ответа сразу
        )

        bot_message = response['message']['content']
        print(f"\n{datetime.now():%H:%M:%S} LLM: {bot_message}\n")
        history.append({'role': 'assistant', 'content': bot_message})


if __name__ == "__main__":
    #без начального промпта
    chat()
```
